```python
# Write a Pyspark query to report the movies with an odd-numbered ID and a description that is not "boring".
# Return the result table in descending order by rating.

data=[(1, 'War','great 3D',8.9)
,(2, 'Science','fiction',8.5) 
,(3, 'irish','boring',6.2)
,(4, 'Ice song','Fantacy',8.6)  
,(5, 'House card','BORING',9.1)]

schema="ID int,movie string,description string,rating double"

df10=spark.createDataFrame(data,schema)

df10.show()
```

```python
#df_movie = df10.filter(when((col("ID")%2==1,True) & when(col("description") == "boring",True)))

df_movie = df10.filter((col("ID")%2==1) & (lower(trim(col("description")))=='boring')).orderBy(col("rating").desc())

df_movie.show()
```

```python
data = [
    ("john", "tomato", 2),
    ("𝚋𝚒𝚕𝚕", "𝚊𝚙𝚙𝚕𝚎", 2),
    ("john", "𝚋𝚊𝚗𝚊𝚗𝚊", 2),
    ("john", "tomato", 3),
    ("𝚋𝚒𝚕𝚕", "𝚝𝚊𝚌𝚘", 2),
    ("𝚋𝚒𝚕𝚕", "𝚊𝚙𝚙𝚕𝚎", 2),
]
schema = "name string,item string,weight int"
df = spark.createDataFrame(data, schema)
#df.show()

df_final = df.groupby(col("name"),col("item")).agg(sum(col("weight")).alias("sum_weight")).orderBy(col("name"))

display(df_final.groupby(col("name")).agg(collect_list(struct(col("item"),col("sum_weight"))).alias("list")))
```

```python
data = [(1, "abc@gmail.com"), (2, "bcd@gmail.com"), (3, "abc@gmail.com")]
schema = "ID int,email string"
df = spark.createDataFrame(data, schema)
#df.show()

df.createOrReplaceTempView("tempdata")

spark.sql("""
select * from tempdata where email in (
select email from tempdata group by email having count(ID) > 1)

""").show()
```

```python
from pyspark.sql.types import *
inputData2 = [('Goa','','AP'),('','AP',None),(None,'','Blr')]
schema2 = StructType([StructField('City1',StringType(),True),
                     StructField('City2',StringType(),True),
                     StructField('City3',StringType(),True)])
df2 = spark.createDataFrame(inputData2,schema2)
```

```python
df2.show()
df3 = df2.withColumn("NewCol",greatest(col("City1"),col("City2"),col("City3")))
df3.show()
```

```python


data=[('John Doe','{"street": "123 Main St", "city": "Anytown"}'),('Jane Smith','{"street": "456 Elm St", "city": "Othertown"}')]
df=spark.createDataFrame(data,schema="name string,address string")
display(df)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
json_schema="street STRING, city STRING"

df3 = df.withColumn("address_json",from_json(col("address"),json_schema))

df3.withColumn("street",col("address_json.street")).withColumn("city",col("address_json.street"))\
.drop(col("address"),col("address_json")).show()
```
