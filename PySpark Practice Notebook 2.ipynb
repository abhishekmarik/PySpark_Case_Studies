```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
# Welcome to your new notebook
# Type here in the cell editor to add code! 
data = [
    (1, "Sagar", 23, "Male", 68.0),
    (2, "Kim", 35, "Female", 90.2),
    (3, "Alex", 40, "Male", 79.1),
]
schema = "Id int,Name string,Age int,Gender string,Marks float"
df = spark.createDataFrame(data, schema)

df.show()
print(df.dtypes)
```

```python

#df_int = spark.createDataFrame([], "dummy int")
list_int,list_str,list_float = [],[],[]
for i in df.dtypes:
    print(i[0],i[1])
    if(i[1]=="int"):
        list_int.append(i[0])
    if(i[1]=="string"):
        list_str.append(i[0])
    if(i[1]=="float"):
        list_float.append(i[0])
        
df_int = df.select(list_int)
df_str = df.select(list_str)
df_float = df.select(list_float)
display(df_int)
display(df_str)
display(df_float)
```

```python
d={}
for col in df.dtypes:
    if col[1] not in d:
        d[col[1]] = [col[0]]
    else:d[col[1]].append(col[0])

print(d)

for key,val in d.items():
    df.select(val).show()
```

```python
data=[
('Rudra','math',79),
('Rudra','eng',60),
('Shivu','math', 68),
('Shivu','eng', 59),
('Anu','math', 65),
('Anu','eng',80)
]
schema="Name string,Sub string,Marks int"
df=spark.createDataFrame(data,schema)
df.show()
```

```python
df.withColumn("Math", when(col("Sub")=="math",col("Marks")))\
  .withColumn("Eng", when(col("Sub")=="eng",col("Marks"))).drop(col("Sub"),col("Marks"))\
  .fillna(0).groupBy(col("Name")).agg(sum(col("Math")).alias("Math"),sum(col("Eng")).alias("Eng")).show()
```

```python
df.orderBy(col("Name"),col("sub")).show()
```

```python
data = [
(1, "A", 1000),
(2, "B", 2000),
(3, "C", 3000),
(4, "D", 4000),
]

schema1 = StructType([
  StructField("ID", IntegerType(), True),
  StructField("Name", StringType(), True),
  StructField("Sal", IntegerType(), True)
])
df2 = spark.createDataFrame(data, schema=schema1)
df2.show()
```

```python
from pyspark.sql.window import Window
windowSpec = Window.orderBy(col("ID")).rowsBetween(Window.unboundedPreceding,Window.currentRow)

df2.withColumn("Sum",sum(col("Sal")).over(windowSpec)).show()
```

```python
schema = StructType([
StructField("id", IntegerType(), nullable=False),
StructField("name", StringType(), nullable=False), 
StructField("age", IntegerType(), nullable=False),
StructField("department", StringType(), nullable=False), 
StructField("salary", DoubleType(), nullable=False)
])
data = [
    (1, "John", 30, "Sales", 50000.0),
    (2, "Alice", 28, "Marketing", 60000.0),
    (3, "Bob", 32, "Finance", 55000.0),
    (4, "Sarah", 29, "Sales", 52000.0),
    (5, "Mike", 31, "Finance", 58000.0)
]
employeeDF = spark.createDataFrame(data, schema)
display(employeeDF)
```

```python
employeeDF.groupBy(col("department")).agg(max("salary").alias("max_sal")).orderBy(col("max_sal")).limit(1)
```

```python
#NULL safe Joins

data1 = [

    (None,"Alice"),
    (2,"Bob"),
    (None, "Charlie"),
    (4,"David")
]

data2 = [

    (None,"X"),
    (2,"Y"),
    (3, "Z"),
    (None,"W")
]

schema1 = "ID Int, Name STRING"
schema2 = "ID Int, Value STRING"

df1 = spark.createDataFrame(data1,schema1)
df2 = spark.createDataFrame(data2,schema2)

df1.show()
df2.show()

#INNER JOIN
df3 = df1.join(df2,df1.ID==df2.ID,"inner")
#df3.show()

#LEFT JOIN
df4 = df1.join(df2,df1.ID==df2.ID,"left")
#df4.show()

#RIGHT JOIN
df5 = df1.join(df2,df1.ID==df2.ID,"right")
#df5.show()

#FULL JOIN
df6 = df1.join(df2,df1.ID==df2.ID,"full")
df6.show()
df7 = df1.join(df2,df1.ID.eqNullSafe(df2.ID),"full")
df7.show()
```
